```python
map_rooms = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["D"],
    "D": ["G"],
    "G": []
}

print("Our map:")
print("          A (Start)")
print("         / \\")
print("        B   C")
print("         \\ /")
print("          D")
print("          |")
print("          G (Treasure 💎)")
print()

print("Doors from each room:", map_rooms)
```

```python
def dfs (start,goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
        room = to_do.pop()
        if room in visited:
            continue
        visited.append(room)
        order.append(room)
        if room == goal:
            return order
        for neighbour in map_rooms[room]:
            to_do.append(neighbour)
    return order
order = dfs("A","G")
print("DFS order:", order)
```
